```python
from bs4 import BeautifulSoup
import requests
```

```python
url = 'https://www.worldometers.info/world-population/population-by-country/'

headers = { 'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'}

page = requests.get(url, headers=headers)

soup = BeautifulSoup(page.text, 'html')
```

```python
table = soup.find_all('table')[0]
print(table)
```

```python
tabletitles = table.find_all('th')
tabletitles
```

```python
Tablehead = [title.text.strip() for title in tabletitles]
Tablehead
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(columns=Tablehead)
df
```

```python
column_data = table.find_all('tr')
```

```python
for row in column_data[1:]:
    row_data = row.find_all('td')
    individual_row_data = []

    for data in row_data:
        text = data.text.strip()
        # Handle multiple Unicode dash variations
        text = text.replace('\u2212', '-')  # Minus sign
        text = text.replace('\u2013', '-')  # En dash
        text = text.replace('\u2014', '-')  # Em dash
        text = text.replace('â', '')        # Remove stray â characters
        text = text.replace('\x88\x92', '-') # Handle your specific encoding

        # Remove extra spaces around negative signs
        text = text.replace('- ', '-').replace(' -', '-')

        individual_row_data.append(text)

    print(individual_row_data)
    length = len(df)
    df.loc[length] = individual_row_data
```
